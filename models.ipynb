```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
data=pd.read_csv("Titanic-Dataset.csv")
```

```python
data.describe()
```

```python
print(data.shape)
data.info()
```

```python
# age/cabin/embarked
data.isnull().sum()
```

```python
# delete imimportant columns

data.drop(["PassengerId","Name"],inplace=True,axis=1)
data.drop("Ticket",axis=1,inplace=True)

```

```python
data.sample(20)
```

```python
len(data["Cabin"].unique())
```

```python
data.drop("Cabin",axis=1,inplace=True)
```

```python
data.head()
```

```python
data.info()
```

```python
data["Age"]=data["Age"].fillna(data["Age"].mean())
```

```python
data.info()
```

```python
data["Embarked"]=data["Embarked"].fillna(data["Embarked"].mode()[0])
```

```python
data.info()
```

```python
data.head()
```

```python
print(data["Embarked"].unique())
print(data["Sex"].unique())
```

```python
encoding_embarked={
    "S":1,
    "C":2,
    "Q":3
}

data["Embarked"] = data["Embarked"].map(
    encoding_embarked
)
```

```python
data.info()
```

```python
sex_encoding={
    "male":1,
    "female":0
}

data["Sex"]=data["Sex"].map(sex_encoding)
```

```python
data.info()
```

```python
y=data["Survived"]
x=data.drop("Survived",axis=1)
```

```python
from sklearn.model_selection import train_test_split

train_x,test_x,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=33)

```

```python
print(train_x.shape)
print(y_train.shape)
print(y_test.shape)
print(test_x.shape)
```

```python
import keras
```

```python
model=keras.Sequential([
    keras.layers.Dense(32,activation="relu"),
    # keras.layers.Dropout(0.4),
    keras.layers.Dense(64,activation="relu"),
    # keras.layers.Dropout(0.4),
    # keras.layers.Dense(16,activation="relu"),
    keras.layers.Dense(1,activation="sigmoid")
])
```

```python
model.compile(
    metrics=["accuracy"], optimizer="adam",loss="binary_crossentropy"
)
```

```python
history=model.fit(
    train_x,y_train,epochs=60,batch_size=4,verbose=1
)
```

```python
model.evaluate(test_x,y_test)
```

```python
model.save("titanic_model.keras")
```

```python
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')

plt.title("Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")

plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')

plt.title("Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")

plt.legend()
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

for key in history.history.keys():
    
    plt.figure(figsize=(6,4))
    
    plt.plot(history.history[key], label=key)
    
    plt.title(f"{key} over epochs")
    plt.xlabel("Epochs")
    plt.ylabel(key)
    
    plt.legend()
    plt.grid(True)
    
    plt.show()
```

```python
data=pd.read_csv("Iris.csv")
```

```python
data.info()
```

```python
data["Species"].unique()
```

```python
map_encoding={
    
    'Iris-setosa':0, 'Iris-versicolor':1, 'Iris-virginica':2
    
}


data["Species"]=data["Species"].map(map_encoding)
```

```python
data.head()
```

```python
data.sample(20)
```

```python
data.drop("Id",inplace=True,axis=1)
data.head()
```

```python
x=data.drop("Species",axis=1)
y=data["Species"]
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=34)
```

```python
model=keras.Sequential([
    keras.layers.Dense(16,activation="relu"),
    keras.layers.Dense(32,activation="relu"),
    keras.layers.Dense(3,activation="softmax")
])

model.compile(
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
    optimizer="adam"
)
```

```python
history=model.fit(
    x_train,
    y_train,
    epochs=60,
)
```

```python
model.evaluate(x_test,y_test)
```

```python
import matplotlib.pyplot as plt

for key in history.history.keys():
    
    plt.figure(figsize=(6,4))
    
    plt.plot(history.history[key], label=key)
    
    plt.title(f"{key} over epochs")
    plt.xlabel("Epochs")
    plt.ylabel(key)
    
    plt.legend()
    plt.grid(True)
    
    plt.show()
```

```python
y_pre=model.predict(x_test)
y_pre[0:5]
```

```python
model.save("iris.keras")
```

```python
data.shape
```

```python
data.drop("id",axis=1,inplace=True)

```

```python
columns=data.columns
for c in columns:
    if "se" in c or "worst" in c:
        data.drop(c,inplace=True,axis=1)
```

```python
data.shape
```

```python
data.sample(5)
```

```python
data["diagnosis"].unique()
```

```python
predict_encoding={
    "B":0,
    "M":1
}

data["diagnosis"]=data["diagnosis"].map(predict_encoding)
```

```python
data.info()
```

```python
y=data["diagnosis"]
x=data.drop("diagnosis",axis=1)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=22)
```

```python
model=keras.Sequential([
    keras.layers.Dense(32,activation="relu"),
    keras.layers.Dense(64,activation="relu"),
    keras.layers.Dense(1,activation="sigmoid")
])

model.compile(
    metrics=["accuracy"], optimizer="adam",loss="binary_crossentropy"
)
```
