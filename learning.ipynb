```python
import numpy as np

y_predicted=np.array([1,1,0,0,1])
y_true= np.array([0.3,0.7,1,0,0.5])
```

```python
def mae(y_predicted,y_true):
    
    total_error=0
    
    for yp,yt in zip(y_predicted,y_true):
        
        total_error+= abs(yp-yt)
        
    
    return total_error / len(y_predicted)


print(mae(y_predicted,y_true))        
    
```

```python
def mse (y_predicted,y_true):
    
    square=(y_predicted-y_true)**2
    
    result=(np.sum(square)/len(square))
    
    return result


yp=np.array([3,2,1])
yt=np.array([4,1,4])

mse(yp,yt)


```

```python
def log_loss(y_true,y_predicted):
    
    epsilon=1e-15
    y_predicted_new=[max(epsilon,x) for x in y_predicted]
    y_predicted_new=[min(x,1-epsilon) for x in y_predicted_new]
    y_predicted_new=np.array(y_predicted_new)
    return - np.mean(y_true*np.log(y_predicted_new)+(1-y_true)*np.log(1-y_predicted_new))

```
